## 1. 인증 준비 및 model.py 다운로드


```python
import os
from pathlib import Path
import urllib.request
from google.colab import userdata

# Colab Secrets: enable notebook access for both names; never print token values.
os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
os.environ["GITHUB_TOKEN"] = userdata.get("GITHUB_TOKEN")

# Download the standalone entry point using authenticated GitHub repository access.
request = urllib.request.Request(
    "https://api.github.com/repos/monkey1soong/Codex-test/contents/Model_test/SmolLM2-360M-Instruct/model.py?ref=main",
    headers={"Authorization": "Bearer " + os.environ["GITHUB_TOKEN"],
             "Accept": "application/vnd.github.raw+json",
             "X-GitHub-Api-Version": "2022-11-28",
             "User-Agent": "SmolLM2-Colab-Test"},
)
with urllib.request.urlopen(request, timeout=30) as response:
    source = response.read()
Path("/content/model.py").write_bytes(source)
del request
os.chdir("/content")
print("준비 완료. 다음 셀의 !python model.py를 실행하세요.")
```
